```python
import numpy as np
import matplotlib.pyplot as plt

# Generate some sample data
np.random.seed(0)
X = np.random.randn(100, 1)
y = 2 * X + 1 + np.random.randn(100, 1) * 0.1

# Plot the data
plt.scatter(X, y, alpha=0.5)
plt.xlabel('X')
plt.ylabel('y')
plt.title('Sample Data')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# একটা simple linear relationship: y = 2x + 1
x = np.linspace(-5, 5, 100) # x এর মান generate করা হলো
y = 2*x + 1

plt.figure(figsize=(6,4)) # figure size নির্ধারণ করা হলো
plt.plot(x, y, label='y = 2x + 1') # linear relationship plot করা হলো
plt.axhline(0, color='gray', linewidth=0.5) # x-axis দেখানো হলো
plt.axvline(0, color='gray', linewidth=0.5) # y-axis দেখানো হলো
plt.scatter([0], [1], color='red', zorder=5)   # y-intercept দেখানো হলো
plt.legend() # legend দেখানো হলো
plt.title("y = mx + c : Linear Regression-এর ভিত্তি") # title দেওয়া হলো
plt.grid(True, alpha=0.3) # grid lines দেখানো হলো
plt.show()
```

## কোডের line-by-line ব্যাখ্যা

1. `import numpy as np`  
   NumPy library import করে এবং ছোট নাম হিসেবে `np` ব্যবহার করে। এই library দিয়ে সংখ্যাভিত্তিক array তৈরি ও হিসাব করা হয়।

2. `import matplotlib.pyplot as plt`  
   Matplotlib-এর `pyplot` module import করে এবং এর সংক্ষিপ্ত নাম দেয় `plt`। এটি graph তৈরি ও দেখানোর কাজে লাগে।

3. `# একটা simple linear relationship: y = 2x + 1`  
   এটি একটি comment। এখানে জানানো হয়েছে যে আমরা `y = 2x + 1` linear equation-এর graph আঁকব। Python comment execute করে না।

4. `x = np.linspace(-5, 5, 100)`  
   `-5` থেকে `5` পর্যন্ত সমান ব্যবধানে মোট `100`টি সংখ্যা তৈরি করে `x` variable-এ রাখে। এই সংখ্যাগুলো graph-এর x-coordinate।

5. `y = 2*x + 1`  
   `x`-এর প্রতিটি মানের জন্য `y = 2x + 1` সূত্র ব্যবহার করে corresponding y-value হিসাব করে। এখানে slope `m = 2` এবং y-intercept `c = 1`।

6. `plt.figure(figsize=(6,4))`  
   নতুন একটি figure তৈরি করে। `figsize=(6,4)` মানে figure-এর width `6` inch এবং height `4` inch।

7. `plt.plot(x, y, label='y = 2x + 1')`  
   `x` ও `y`-এর মান ব্যবহার করে সরলরেখাটি আঁকে। `label`-এর text legend-এ দেখানো হয়।

8. `plt.axhline(0, color='gray', linewidth=0.5)`  
   `y = 0` অবস্থানে একটি horizontal ধূসর রেখা আঁকে। এটিই graph-এর **x-axis**। `linewidth=0.5` রেখাটিকে পাতলা রাখে।

9. `plt.axvline(0, color='gray', linewidth=0.5)`  
   `x = 0` অবস্থানে একটি vertical ধূসর রেখা আঁকে। এটিই graph-এর **y-axis**।

10. `plt.scatter([0], [1], color='red', zorder=5)`  
    `(0, 1)` coordinate-এ একটি লাল বিন্দু বসায়। এটি রেখাটির y-intercept, কারণ `x = 0` হলে `y = 1`। `zorder=5` বিন্দুটিকে রেখা ও grid-এর উপরে দৃশ্যমান রাখে।

11. `plt.legend()`  
    `plt.plot()`-এ দেওয়া label ব্যবহার করে graph-এর legend দেখায়।

12. `plt.title("y = mx + c : Linear Regression-এর ভিত্তি")`  
    Graph-এর উপরে একটি title যোগ করে। সাধারণ linear equation হলো `y = mx + c`, যেখানে `m` slope এবং `c` y-intercept।

13. `plt.grid(True, alpha=0.3)`  
    Graph-এর grid line চালু করে। `alpha=0.3` grid-কে হালকা বা 30% opaque রাখে।

14. `plt.show()`  
    তৈরি করা সম্পূর্ণ graph output হিসেবে দেখায়।

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100) # x এর মান generate করা হলো
slopes = {'m=2': 2, 'm=0.5': 0.5, 'm=0': 0, 'm=-1': -1}

plt.figure(figsize=(6,4))
for label, m in slopes.items():
    plt.plot(x, m*x, label=label)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.title("বিভিন্ন Slope-এর লাইন")
plt.grid(True, alpha=0.3)
plt.show()

# দুইটা বিন্দু থেকে slope বের করা
def calculate_slope(p1, p2):
    return (p2[1] - p1[1]) / (p2[0] - p1[0])

print(calculate_slope((-4, 2), (3, -2)))   # -0.571... — negative slope
```

## বিভিন্ন slope-এর কোড: line-by-line ব্যাখ্যা

1. `import numpy as np`  
   NumPy library import করে এবং সংক্ষিপ্ত নাম `np` দেয়। সংখ্যার array তৈরি ও হিসাব করতে এটি ব্যবহার করা হয়।

2. `import matplotlib.pyplot as plt`  
   Matplotlib-এর `pyplot` module import করে এবং সংক্ষিপ্ত নাম `plt` দেয়। এটি graph আঁকতে ব্যবহৃত হয়।

3. `x = np.linspace(-5, 5, 100)`  
   `-5` থেকে `5` পর্যন্ত সমান ব্যবধানে মোট `100`টি x-value তৈরি করে।

4. `slopes = {'m=2': 2, 'm=0.5': 0.5, 'm=0': 0, 'm=-1': -1}`  
   একটি dictionary তৈরি করে। প্রতিটি key graph-এর label এবং সংশ্লিষ্ট value হলো slope `m`। এখানে চারটি slope রাখা হয়েছে: `2`, `0.5`, `0` ও `-1`।

5. `plt.figure(figsize=(6,4))`  
   width `6` inch ও height `4` inch-এর একটি নতুন figure তৈরি করে।

6. `for label, m in slopes.items():`  
   Dictionary-এর প্রতিটি key-value pair-এর উপর loop চালায়। প্রতিবার key যায় `label`-এ এবং slope-এর সংখ্যা যায় `m`-এ।

7. `plt.plot(x, m*x, label=label)`  
   প্রতিটি slope-এর জন্য `y = mx` হিসাব করে একটি line আঁকে। সব line origin `(0, 0)` দিয়ে যায়, কারণ এখানে y-intercept বা `c = 0`। `label` legend-এ line-টির পরিচয় দেখায়।

8. `plt.axhline(0, color='gray', linewidth=0.5)`  
   `y = 0` অবস্থানে পাতলা horizontal ধূসর line আঁকে—এটি x-axis।

9. `plt.axvline(0, color='gray', linewidth=0.5)`  
   `x = 0` অবস্থানে পাতলা vertical ধূসর line আঁকে—এটি y-axis।

10. `plt.legend()`  
    প্রতিটি line-এর label (`m=2`, `m=0.5` ইত্যাদি) দিয়ে legend দেখায়।

11. `plt.title("বিভিন্ন Slope-এর লাইন")`  
    Graph-এর উপরে title যোগ করে।

12. `plt.grid(True, alpha=0.3)`  
    হালকা grid line দেখায়। `alpha=0.3` মানে grid 30% opaque।

13. `plt.show()`  
    সম্পূর্ণ graph output-এ দেখায়।

14. `# দুইটা বিন্দু থেকে slope বের করা`  
    এটি একটি comment। নিচের function-এর উদ্দেশ্য জানায়; Python এটি execute করে না।

15. `def calculate_slope(p1, p2):`  
    `calculate_slope` নামে একটি function তৈরি করে। এটি দুটি point গ্রহণ করে: `p1 = (x₁, y₁)` এবং `p2 = (x₂, y₂)`।

16. `return (p2[1] - p1[1]) / (p2[0] - p1[0])`  
    Slope-এর সূত্র `m = (y₂ - y₁) / (x₂ - x₁)` ব্যবহার করে ফল return করে। Tuple-এর index `0` হলো x-value এবং index `1` হলো y-value।

17. `print(calculate_slope((-4, 2), (3, -2)))`  
    Point দুটি `(-4, 2)` ও `(3, -2)` function-এ পাঠিয়ে slope print করে। হিসাব: `(-2 - 2) / (3 - (-4)) = -4/7 ≈ -0.5714`। ফল negative হওয়ায় x বাড়লে line নিচের দিকে নামে।

### Slope দেখে line বোঝা

- `m = 2`: line দ্রুত উপরের দিকে ওঠে—steep positive slope।
- `m = 0.5`: line ধীরে উপরের দিকে ওঠে—gentle positive slope।
- `m = 0`: line horizontal থাকে; x বদলালেও y বদলায় না।
- `m = -1`: x বাড়লে line নিচের দিকে নামে—negative slope।

> **সতর্কতা:** দুই point-এর x-value একই হলে `x₂ - x₁ = 0` হবে। তখন division by zero ঘটবে এবং vertical line-এর slope undefined হবে।
