```python
# string to integer conversion
def string_to_integer(s):
    try:
        return int(s)
    except ValueError:
        return None

print(string_to_integer("123"))  # Output: 123
```

```python
age = "28"
age = int(age)  # Convert string to integer
print(age)  # Output: 28
```

```python
# integer to string conversion

def integer_to_string(n):
    try:
        return str(n)
    except ValueError:
        return None

print(integer_to_string(123))  # Output: "123"
```

```python
age = 29
# message = f"I am {age} years old."  # Convert integer to string and concatenate
message = "I am " + str(age) + " years old."  # Convert integer to string and concatenate
print(message)  # Output: "I am 29 years old."
```

```python
# integer to float and decimal conversion
from decimal import Decimal


def integer_to_float(n):
    try:
        return float(n)
    except ValueError:
        return None

print(integer_to_float(123))  # Output: 123.0

def integer_to_decimal(n):
    try:
        return Decimal(n) # Convert integer to decimal
    except ValueError:
        return None

print(integer_to_decimal(123.30))  # Output: 123
```
