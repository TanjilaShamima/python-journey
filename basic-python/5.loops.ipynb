```python
numbers = input("Enter a list of numbers separated by spaces: ").split()
numbers = [int(num) for num in numbers]  # Convert input strings to integers

max_num = max(numbers)
min_num = min(numbers)

for number in numbers:
    if number == max_num or number == min_num:
        if number == max_num:
            print(f"{number} is the maximum number.")
            if number%2 == 0:
                print(f"{number} is an even number.")
            else:
                print(f"{number} is an odd number.")
        else:
            print(f"{number} is the minimum number.")
            if number%2 == 0:
                print(f"{number} is an even number.")
            else:
                print(f"{number} is an odd number.")
    else:
        print(f"{number} is not the maximum or minimum number.")
```

```python
# range function = range(start, stop, step) = value >= start and value < stop and value = start + step * n where n is a non-negative integer

for i in range(1, 30, 2):
    print(i)
```

```python
# reverse range
for i in range(10, 0, -3):
    print(i)
```

```python
# while loop

count = 0
while count < 5:
    print(count)
    count += 1
```

```python
# break - continue - pass

numbers = input("Enter a number: ").split()
numbers = [int(num) for num in numbers]

min_num = min(numbers)
max_num = max(numbers)

print("Minimum num", min_num)
print("Maximum num", max_num)

for number in numbers:
    if number == min_num:
        print(f"We got the minimum number: {number}")
        continue
    elif number == max_num:
        print(f"We got the maximum number: {number}")
        break
    else:
        print(f"{number} is neither the maximum nor the minimum number.")
```
