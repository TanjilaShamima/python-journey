```python
# lambda

number = int(input("Enter a number:"))

def square_function(x):
    return x ** 2

square = lambda x: x * x

result = square(number)
result_function = square_function(number)

print(f"The square of {number} is {result}")
print(f"The square of {number} using function is {result_function}")
```

```python
# two parameters

numbers = input("Enter two numbers separated by space:").split()
num1 = int(numbers[0])
num2 = int(numbers[1])

add = lambda a, b: a + b
subtract = lambda a, b: a - b
multiply = lambda a, b : a * b
division = lambda a, b : a / b if b != 0 else "Division by zero is not allowed"

print(f"The sum of {num1} and {num2} is {add(num1, num2)}")
print(f"The difference of {num1} and {num2} is {subtract(num1, num2)}")
print(f"The product of {num1} and {num2} is {multiply(num1, num2)}")
print(f"The division of {num1} by {num2} is {round(division(num1, num2), 2)}")
```

```python
# check even or odd

number = int(input("Enter a number to check if it is even or odd:"))

check_even_or_odd = lambda x : "Even Number" if x%2 == 0 else "Odd Number"

print(f"The number {number} is an {check_even_or_odd(number)}")
```
