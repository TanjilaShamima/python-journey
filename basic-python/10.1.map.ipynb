```python
# map()

numbers = [1, 2, 3, 4, 5]

squared_numbers = map(lambda x : x**2, numbers)

result = list(squared_numbers)

print(f"The squared numbers are: {result}")
```

```python
numbers = input("Enter numbers separated by commas: ").split(' ')
numbers1 = [int(x) for x in numbers]
numbers2 = list(map(lambda x: int(x), numbers))
numbers3 = list(map(int, numbers))

print(f"The numbers are: {numbers1}")
print(f"The numbers are: {numbers2}")
print(f"The numbers are: {numbers3}")
```
