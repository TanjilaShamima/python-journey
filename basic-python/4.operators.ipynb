```python
# Arithmetic Operators

a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a // b)  # Floor division
print(a % b)   # Remainder
print(a ** b)  # Power
```

```python
# Comparison Operators

a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# Logical Operators
a = True
b = False

print(a and b)  # AND
print(a or b)   # OR
print(not a)    # NOT
```

```python
age = int(input("Enter your age: "))
age >= 18 and (has_permission := input("Do you have permission? (yes/no): ").strip().lower() == "yes")  # Convert input to boolean

if age > 18:
    if has_permission:
        print("Access granted.")
    else:
        print("Access denied. Permission required.")
elif age == 18:
    if has_permission:
        print("Access granted with caution.")
    else:
        print("Access denied. Permission required.")
else:
    print("Access denied.")
```

```python
# Ternary Expression

age = int(input("Enter your age: "))

status = "Adult" if age >= 18 else "Minor"
# in JS 
# const status = age >= 18 ? "Adult" : "Minor";
print(status)
```
