```python
# Mini Project: Student Result Analysis

import numpy as np

marks = np.array([
    [85, 90, 78], 
    [92, 88, 95], 
    [80, 85, 82],
    [78, 80, 85]
])

students = ["Alice", "Bob", "Charlie", "David"]
subjects = ["Math", "Science", "English"]

# Calculate total marks for each student
total_marks = np.sum(marks, axis=1)

print("Total Marks for each student:")
for i in range(len(students)):
    print(f"{students[i]}: {total_marks[i]}")

# Calculate average marks for each subject
average_marks = np.mean(marks, axis=0)
print("\nAverage Marks for each subject:")
for i in range(len(subjects)):
    print(f"{subjects[i]}: {average_marks[i]:.2f}")

# Find the student with the highest total marks
highest_total_index = np.argmax(total_marks)
print(f"\nStudent with the highest total marks: {students[highest_total_index]}")

# Find the subject with the lowest average marks
lowest_average_index = np.argmin(average_marks)
print(f"Subject with the lowest average marks: {subjects[lowest_average_index]}")

# pass of fail based on average marks
pass_fail = np.where(average_marks >= 80, "Pass", "Fail")
print("\nPass/Fail status for each subject:")
for i in range(len(subjects)):
    print(f"{subjects[i]}: {pass_fail[i]}")

# percentage of marks for each student
percentage = (total_marks / 300) * 100
print("\nPercentage of marks for each student:")
for i in range(len(students)):
    print(f"{students[i]}: {percentage[i]:.2f}%")
```
