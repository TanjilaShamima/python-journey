# Important comparison operators

Operator	Meaning	Example

==	equal	x == 5

!=	not equal	x != 5

>	greater than	x > 5

<	less than	x < 5

>=	greater than or equal	x >= 5

<=	less than or equal	x <= 5

# Logical operators
Operator	Meaning

and	 = both conditions must be true

or	 = at least one condition must be true

not	 = reverses the condition

```python
# List Comprehension
squares = [x**2 for x in range(10)]
print("Squares:", squares)
```

```python
numbers = [10, 23, 30, 40, 50, 51]
square = [x**2 for x in numbers if x%5 == 0]
print("Squares of numbers:", square)
```

```python
# List
fruits = ["apple", "banana", "cherry"]
print("Fruits:", fruits)
students = ["Alice", "Bob", "Charlie"]
print("Students:", students)
mixed_list = [1, "hello", 3.14, True]
print("Mixed List:", mixed_list)

# Accessing List Elements
print("First Fruit:", fruits[0])

# Negative Indexing
print("Last Fruit:", fruits[-1])

# Slicing
# structure: list[start:end] (end is exclusive)
print("First Two Fruits:", fruits[0:2])
print("Last Two Fruits:", fruits[-2:])
```

```python
## Adding Elements to a List
fruits.append("orange")
print("Fruits after append:", fruits)

fruits.insert(1, "grape")
print("Fruits after insert:", fruits)
```

```python
## Remove Elements from a List
fruits.remove("grape")
print("Fruits after remove:", fruits)
```

```python
# ML Examples
prices = [100, 150, 200, 250, 300]
# after 10% discount
discount_prices = [price * 0.1 for price in prices]
prices_after_discount = [price * 0.9 for price in prices]
print("Discount Prices:", discount_prices)
print("Prices after 10% discount:", prices_after_discount)

# 15% discount if price is greater than or equal to 200
discount_prices = [price * 0.15 for price in prices if price >= 200]
print("Discount Prices (15% off prices >= 200):", discount_prices)
prices_after_discount = [price * 0.85 for price in prices if price >= 200]
print("Prices after 15% discount (prices >= 200):", prices_after_discount)
```

```python
# Dictionary
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A",
    "is_passed": True
}

# name is key and Alice is value
print("Student Dictionary:", student)
```

```python
# accessing values using keys
print("Student Name:", student["name"])
print("Student Age using get():", student.get("age"))
print("Student Age:", student["age"])
print("Student Grade:", student["grade"])
print("Student Passed:", student["is_passed"])
```
