```python
values = input("Enter numbers separated by space: ")
numbers = [int(x) for x in values.split()]

def calculate_sum(nums):
    return sum(nums)

def calculate_average(nums):
    return sum(nums) / len(nums)

print("Sum of numbers:", calculate_sum(numbers))
print("Average of numbers:", calculate_average(numbers))
```

```python
def calculate_area(length, width):
    return length * width

input_data = input("Enter length and width separated by space: ")
# arr = input_data.split()
# length = float(arr[0])
# width = float(arr[1])
length, width = map(float, input_data.split())
print("Area of rectangle:", calculate_area(length, width))
```

```python
# Lambda, map and apply

def square(x):
    return x**2
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(square, numbers))

print("Squared numbers:", squared_numbers)

# using lambda function
squared_numbers_lambda = list(map(lambda x: x**2, numbers))
print("Squared numbers (lambda):", squared_numbers_lambda)

```

```python
# apply() 
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# filtering rows where Age > 28
filtered_df = df[df['Age'] > 28]
print("Filtered DataFrame:")
print(filtered_df)
```
