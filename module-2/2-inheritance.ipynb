```python
# Inheritance
# Inheritance is the concept of creating a new class that is based on an existing class. The new class inherits the attributes and methods of the existing class, and can also have its own attributes and methods.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduction(self):
        return f"My name is {self.name} and I am {self.age} years old."

class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)
        self.grade = grade

    def student_introduction(self):
        return f"{self.introduction()} and I am in grade {self.grade}."
    
student1 = Student("Alice", 20, "A")
print(student1.student_introduction())
```

Types of Inheritance

1. Single Inheritance: In single inheritance, a class (called thechild class) inherits from a single parent class. The child class can access the properties and methods of the parent class.

2. Multiple Inheritance: In multiple inheritance, a class can inherit from more than one parent class. This allows the child class to access properties and methods from multiple parent classes.

3. Multilevel Inheritance: In multilevel inheritance, a class inherits from a parent class, which in turn inherits from another parent class. This creates a chain of inheritance.

```python
# Single Inheritance: In single inheritance, a class (called thechild class) inherits from a single parent class. The child class can access the properties and methods of the parent class.

class Employee(Person):
    def __init__(self, name, age, position):
        super().__init__(name, age)
        self.position = position

    def employee_introduction(self):
        return f"{self.introduction()} and I work as a {self.position}."
    
employee1 = Employee("Bob", 30, "Software Engineer")
print(employee1.employee_introduction())
```

```python
# Multilevel Inheritance: In multilevel inheritance, a class inherits from a parent class, which in turn inherits from another parent class. This creates a chain of inheritance.

class Animal:
    def __init__(self, color):
        self.color = color
    def eat(self):
        return "Animal is eating. Its color is " + self.color + "."

class Dog(Animal):
    def __init__(self, color, breed):
        super().__init__(color)
        self.breed = breed
    def bark(self):
        return "Dog is barking. Its breed is " + self.breed + "."
    
class Puppy(Dog):
    def __init__(self, color, breed):
        super().__init__(color, breed)
    def weep(self):
        return "Puppy is weeping. Its breed is " + self.breed + "."
    
puppy1 = Puppy("Brown", "Golden Retriever")
print(puppy1.eat())
print(puppy1.bark())
print(puppy1.weep())
```

```python
# Multiple Inheritance
class Flyer:
    def fly(self):
        return "I can fly."
class Swimmer:
    def swim(self):
        return "I can swim."
class Duck(Flyer, Swimmer):
    def quack(self):
        return "I can quack."
duck1 = Duck()
print(duck1.fly())
print(duck1.swim())
print(duck1.quack())
```
