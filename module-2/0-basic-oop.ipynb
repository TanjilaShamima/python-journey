```python
class Student:
    def __init__(self, name, age, department):
        self.name = name
        self.age = age
        self.department = department
    
    # here introduction is a method
    def introduction(self):
        return f"My name is {self.name}, I am {self.age} years old and I study {self.department}."


student1 = Student("Alice", 20, "Computer Science")
student1.name = "Bob"
print(student1.introduction())
```

```python
class Student:
    def __init__(self, name, age, department, score):
        self.name = name
        self.age = age
        self.department = department
        self.score = score
    
    def introduction(self):
        return f"My name is {self.name}, I am {self.age} years old and I study {self.department}."

    def is_passed(self):
        if self.score >= 60:
            return "Passed"
        else:
            return "Failed"
        
student1 = Student("Alice", 20, "Computer Science", 85)
print(student1.introduction())
print("The student has", student1.is_passed())
```

```python
# Constructor Example with Real Life

class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        return f"{self.account_holder} deposited {amount}. New balance is {self.balance}."
    
    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds."
        else:
            self.balance -= amount
            return f"{self.account_holder} withdrew {amount}. New balance is {self.balance}."
account1 = BankAccount("John Doe", 1000)
print(account1.deposit(500))
print(account1.withdraw(200))
```
