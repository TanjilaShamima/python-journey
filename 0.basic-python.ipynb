```python
# Indexing

text = "Hello, World!"
print(text[0])  # Output: H
print(text[7])  # Output: W
print(text[-1])  # Output: !
```

```python
# slicing = text[start:stop:step] = value >= start and value < stop

numbers = [10, 1, 2, 3, 4, 5, 6, 7, 8, 9]

updated_number = numbers[0:5:2]
updated_number2 = numbers[0:5:1]  # This will take elements from index 0 to 4 with a step of 2

print(updated_number)  # Output: [10, 2, 4]
print(updated_number2)  # Output: [10, 1, 2, 3, 4]
```

```python
# string methods

text = "  Python Programming  "

print("lower", text.lower())
print("upper", text.upper())
print("title", text.title())
print("capitalize", text.capitalize())  # capitalizes the first letter of the string
print("strip", text.strip())  # strips whitespace from both ends
print("replace", text.replace("Python", "Java"))
print("find", text.find("Programming"))
print("count", text.count("m"))
```

```python
# split and join

sentence = "Python is easy"

words = sentence.split()  # splits the string into a list of words

print(words)  # Output: ['Python', 'is', 'easy']

print("-".join(words))  # Output: Python-is-easy
```
