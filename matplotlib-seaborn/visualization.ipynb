```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
df = pd.read_csv("../files/pandas_A_to_Z_practice_dataset.csv")
print("head of the dataset:")
df.head(3)
```

```python
print("info of the dataset:")
df.info()
print("describe of the dataset:")
df.describe()
```

```python
sns.set_theme(style="whitegrid")
```

```python

df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar="False")
plt.title("Missing Values Heatmap")
plt.show()
```

```python
df["Age"]  = df["Age"].fillna(df["Age"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Region"] = df["Region"].fillna(df["Region"].mode()[0])
df["Last_Promotion_Date"] = df["Last_Promotion_Date"].fillna(df["Last_Promotion_Date"].mode()[0])
```

```python
df.isna().sum()
```

```python
df["Salary_BDT"] = df["Salary_BDT"].fillna(df["Salary_BDT"].mean())
df["Manager_ID"] = df["Manager_ID"].fillna(df["Manager_ID"].mode()[0])
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Salary_BDT", bins=20, kde=True, edgecolor="black", color="skyblue")
plt.title("Distribution of Salary")
plt.xlabel("Salary (BDT)")
plt.ylabel("Frequency")
plt.xticks(
    range(20000, 150001, 10000),
    rotation=45
)
plt.show()
```

```python
df
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Salary_BDT", bins=20, kde=True)
plt.title("Distribution of Salary")
plt.xlabel("Salary (BDT)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x="Salary_BDT", fill=True, color="skyblue")
plt.title("Kernel Density Estimation of Salary")
plt.xlabel("Salary (BDT)")
plt.ylabel("Density")
plt.xticks(
    range(20000, 150001, 10000),
    rotation=45
)
plt.show()
```

```python
# Boxplot

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Salary_BDT", color="lightgreen")
plt.title("Boxplot of Salary")
plt.xlabel("Salary (BDT)")
# plt.xticks(
#     range(20000, 150001, 10000),
#     rotation=45
# )
plt.show()
```

```python
# Barplot
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="Department", y="Salary_BDT", ci=None, palette="pastel")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (BDT)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Countplot
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Department", palette="pastel")
plt.title("Count of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
df["Final_Salary_BDT"] = df["Salary_BDT"] + (df["Salary_BDT"] * 0.1)  # Assuming a 10% increase for final salary

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Salary_BDT", y="Final_Salary_BDT")
plt.title("Final Salary vs Final Score")
plt.xlabel("Final Salary (BDT)")
plt.ylabel("Final Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Experience_Years", y="Final_Salary_BDT", scatter_kws={"color": "blue"}, line_kws={"color": "red"})
plt.title("Experience Years vs Final Salary with Regression Line")
plt.xlabel("Experience Years")
plt.ylabel("Final Salary (BDT)")
plt.show()
```
