## Plot Selection Map — কখন কোন plot?
| Purpose | Best Plot | Use Case |
|---|---|---|
| Trend দেখতে | Line plot | Month-wise sales, year-wise growth |
| Category compare করতে | Bar plot | Department-wise average score |
| Distribution দেখতে | Histogram | Students’ score distribution |
| Density দেখতে | KDE plot | Smooth distribution |
| Outlier দেখতে | Box plot | Salary, score, age outliers |
| Relationship দেখতে | Scatter plot | Study hours vs score |
| Correlation দেখতে | Heatmap | Numeric columns correlation |
| Category + distribution দেখতে | Violin plot | Score spread by department |
| Count দেখতে | Count plot | Male/female count, department count |
| Multiple relationships দেখতে | Pair plot | Numeric feature relationships |
| Percentage/share দেখতে | Pie chart | Market share, category share |
| Two numeric + category দেখতে | Scatter plot with hue | Score vs hours by department |

## Line Plot
কখন use করবে?

যখন data-এর মধ্যে time/order/trend আছে।

Examples:

Month-wise sales
Year-wise population
Daily temperature
Training loss over epochs
Stock price movement

```python
import matplotlib.pyplot as plt
import seaborn as sns

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [200, 250, 300, 280, 350]

df = pd.DataFrame({
    'Month': months,
    'Sales': sales
})

plt.plot(df['Month'], df['Sales'], marker='o')
sns.barplot(data=df, x='Month', y='Sales', alpha=0.5)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.savefig("sales_trend.png")
plt.show()
```

```python
sns.lineplot(data=df, x="Month", y="Sales")
plt.title("Monthly Sales Trend")
plt.show()
```

```python
# BarPlot

departments = ['HR', 'Finance', 'IT', 'Marketing']
avg_salaries = [50000, 60000, 70000, 65000]

df = pd.DataFrame({
    'departments': departments,
    'avg_salaries': avg_salaries
})

plt.bar(departments, avg_salaries, color='blue', alpha=0.7)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.grid(axis='x')
plt.savefig("bar_plot.png")
plt.show()
```

```python
# using seaborn

sns.barplot(data=df, x='departments', y='avg_salaries', palette='viridis')
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.grid(axis='x')
plt.savefig("bar_plot_seaborn.png")
plt.show()
```

Count Plot
কখন use করবে?

যখন শুধু count/frequency দেখতে হবে।

Examples:

কতজন student কোন department-এ আছে
কতজন male/female
কতজন pass/fail
কোন category কতবার এসেছে

```python
df = pd.DataFrame({
    'departments': ['HR', 'Finance', 'IT', 'Marketing'],
    'avg_salaries': [50000, 60000, 70000, 65000]
})

counts = df['departments'].value_counts()

plt.bar(counts.index, counts.values, color='orange', alpha=0.7)
plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.grid(axis='y')
plt.savefig("employee_count_bar_plot.png")
plt.show()
```
