```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 45],
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]
}

df = pd.DataFrame(data)
df
```

```python
print("df.head(3)", df.head(3))  # Display the first few rows of the DataFrame
print("df.describe()", df.describe())  # Get summary statistics of the DataFrame
print("df.info()", df.info())  # Get information about the DataFrame, including data types and non-null counts
print("df.shape", df.shape)  # Get the shape of the DataFrame (rows, columns)
print("df.columns", df.columns)  # Get the column names of the DataFrame
print("df['Name']", df['Name'])  # Access a specific column by name
print("df.index", df.index)  # Get the index of the DataFrame
print("df.dtypes", df.dtypes)  # Get the data types of each column in the DataFrame
```
