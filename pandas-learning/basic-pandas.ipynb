```python
scores = pd.Series([85, 90, 75, 68, 92], index=['Alice', 'Bob', 'Charlie', 'David', 'Eva'])

scores

print("Index:", scores.index)
print("Values:", scores.values)
print(scores['Bob'])  # Accessing value by index label
print(scores.iloc[0])  # Accessing value by integer location
```

```python
print(scores.index)
print(scores.values)
print(scores.dtype)
print(scores.shape)
print(scores.size)
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 45],
    'Score': [85, 90, 75, 68, 92],
    'Status': ['Pass', 'Pass', 'Pass', 'Fail', 'Pass']
}

df = pd.DataFrame(data)

df
```

```python
print(df.head(3))  # Display the first 5 rows
print(df.shape)  # Display the shape of the DataFrame
print(df.index)  # Display the index of the DataFrame
print(df.dtypes)  # Display the data types of each column
print(df.size)  # Display the total number of elements in the DataFrame
print(df.ndim)  # Display the number of dimensions of the DataFrame
```
