```python
df.rename(columns={"Full_Name": "Name"}, inplace=True)

df
```

```python
df["Bonus"] = df["Sales_BDT"] * 0.07  # Create a new column 'Bonus' based on the 'Sales_BDT' column
df
```

```python
df["Total_Salary"] = df["Salary_BDT"] + df["Bonus"]  # Create a new column 'Total_Salary' based on the 'Salary_BDT' and 'Bonus' columns
df
```

```python
df["Status"] = df["Age"].apply(lambda x: "Adult" if x > 18 else "Minor")  # Create a new column 'Status' based on the 'Age' column
df.sample(5)  # Display a random sample of 5 rows from the DataFrame
```

```python
df.sort_values(by="Total_Salary", ascending=False)  # Sort the DataFrame by the 'Total_Salary' column in descending order
```
