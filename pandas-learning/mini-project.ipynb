```python
import pandas as pd

data = {
    "Name": ["Tanjila", "Rafi", "Mina", "Sakib", "Nila"],
    "Department": ["CSE", "EEE", "CSE", "BBA", "EEE"],
    "Marks": [85, 70, None, 90, 60],
    "City": ["Dhaka", "Sylhet", "Dhaka", "Khulna", None]
}

df = pd.DataFrame(data)

print(df.head())
print(df.info())
print(df.isnull().sum())

df["Marks"].fillna(df["Marks"].mean(), inplace=True)
df["City"].fillna(df["City"].mode()[0], inplace=True)

df["Grade"] = df["Marks"].apply(
    lambda mark: "A" if mark >= 80 else "B" if mark >= 70 else "C"
)

print(df.groupby("Department")["Marks"].mean())
print(df["City"].value_counts())

df.to_csv("student_result.csv", index=False)

print(df)
```
