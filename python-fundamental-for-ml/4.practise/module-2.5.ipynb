 Write a function introduce(name, age) that prints:

 My name is <name> and I am <age> years old.
 
Call it using positional arguments.

Sample Input:
Rhim 28
Sample Output:
My name is Rahim and I am 28 years old.


```python
def introduction(name, age):
    return f"My name is {name} and I am {age} years old"

input_data = input("Enter your name and age with space: ")
name, age = map(str, input_data.split())
print(introduction(name, age))
```

Write a Python generator function called batch_generator. It should take a list of numbers (representing data samples) and a batch_size. The generator should yield one batch at a time as a list.

Example Input:
data = [1, 2, 3, 4, 5, 6, 7, 8]
batch_size = 3

Expected Output (when iterated):
[1, 2, 3]
[4, 5, 6]
[7, 8]


```python
def batch_generator(data, batch_size):
    if batch_size <= 0:
        raise ValueError("batch_size must be a positive integer")

    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]


input_data = input("Enter numbers separated by spaces: ")
input_batch_size = int(input("Enter batch size: "))
data = input_data.split()

for batch in batch_generator(data, input_batch_size):
    print(batch)
```

Given a list [1,2,3,4], use map() and lambda to create a new list with squares of each number.

Sample Input:
 No input

Sample Output:
[1, 4, 9, 16]


```python
input_data = input("Enter your list ")

def make_squares(numbers):
    new_list = list(map(lambda x: x ** 2, numbers))
    return new_list 

numbers = list(map(int, input_data.split()))
squared_numbers = make_squares(numbers)
print(squared_numbers)
```

 Write a function power(base, exponent) that returns the result of base raised to exponent. If the exponent is not given, it should calculate the square.
Sample Input:
5
Sample Output:
25
Sample Input 2:
5 3
Sample Output 2:
125


```python
input_data = input("Enter your items separated by spaces: ")
input_list = list(map(int, input_data.split()))

def power(base, exponent):
    if not exponent:
        return base**2
    return base ** exponent

base = input_list[0]
exponent = input_list[1] if len(input_list) > 1 else None

result = power(base, exponent)
print(f"{base} raised to the power of {exponent} is: {result}")
```

Given a list [1,2,3,4,5,6], use filter() to select even numbers and map() with lambda to square them.
Sample Input:
 No input
Sample Output:
[4, 16, 36]


```python
input_data = input("Enter your list ")
input_list = list(map(int, input_data.split()))

def make_squares(numbers):
    new_list = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, numbers)))
    return new_list

squared_evens = make_squares(input_list)
print("Inputs are", input_list)
print(squared_evens)
```
