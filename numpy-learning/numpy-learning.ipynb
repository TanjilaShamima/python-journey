```python
import numpy as np
array = [2, 3, 4, 5, 6]

numpy_array = np.array(array)

print(array)
print(numpy_array)

print(type(array))
print(type(numpy_array))
```

```python
numbers = [1, 2, 3, 4, 5]

print(numbers * 2)

print(np.array(numbers) * 2)
```

```python
# Dimensions

scalar = np.array(10) # 0D array
vector = np.array([1, 2, 3, 4, 5]) # 1D array
matrix = np.array(
    [
        [1, 2, 3], 
        [4, 5, 6]
    ]
) # 2D array
tensor = np.array(
    [
        [
            [1, 2], 
            [3, 4]
        ], 
        [
            [5, 6], 
            [7, 8]
        ]
    ]
) # 3D array

print("Scalar:", scalar)
print("Vector:", vector)
print("Matrix:", matrix)
print("Tensor:", tensor)
```

```python
# important attributes
array = np.array([[1, 2, 3], [4, 5, 6]])
print("Array:\n", array)
print("Shape:", array.shape)
print("Number of dimensions:", array.ndim)
print("Size:", array.size)
print("Data type:", array.dtype)
print("Item size (in bytes):", array.itemsize)
print("Total bytes:", array.nbytes)

print("Number of rows:", array.shape[0])
print("Number of columns:", array.shape[1])
```

```python
array = np.array([[1, 2, 3], [4, 5, 6]], dtype = np.float32)
print("Array:\n", array)
print(array.dtype)
```

```python
float_array = np.array([1.2, 2.8, 3.5])

integer_array = float_array.astype(np.int32)

print(integer_array)
```

```python
# array creation

array = np.array([1, 2, 3, 4, 5])
print("Array:", array)
zeros_array = np.zeros((2, 3))
print("Zeros Array:\n", zeros_array)

ones_array = np.ones((2, 3))
print("Ones Array:\n", ones_array)

filled = np.full((2, 3), 5)


print("Filled Array:\n", filled)

empty_array = np.empty((2, 3))
print("Empty Array:\n", empty_array)

identity = np.eye(3)
print("Identity Matrix:\n", identity)

diagonal = np.diag([10, 20, 30])
print("Diagonal Matrix:\n", diagonal)

arrange = np.arange(0, 10, 2)
print("Arange Array:", arrange)

linespace = np.linspace(0, 1, 5)
print("Linspace Array:", linespace)

# difference between arrange and linspace is that arange creates an array with a specified step size, while linspace creates an array with a specified number of evenly spaced values between two endpoints.

logspace = np.logspace(0, 2, 5)
print("Logspace Array:", logspace)
```

```python
# Random numbers

rng = np.random.default_rng(seed=10)  # Create a random number generator with a seed for reproducibility
print("Random numbers from uniform distribution:", rng.random(7))

random_integers = rng.integers(low=1, high=10, size=5)
print("Random integers:", random_integers)

random_floats = rng.uniform(low=0, high=1, size=5)
print("Random floats from uniform distribution:", random_floats)

normal_distribution = rng.normal(loc=0, scale=1, size=5) # 
print("Random numbers from normal distribution:", normal_distribution)

values = rng.uniform(
    low=0,
    high=10,
    size=5
)
print("Random numbers from uniform distribution:", values)
```

```python
values = rng.normal(
    loc=0,
    scale=1,
    size=10
)

print("Random numbers from normal distribution:", values)
```

```python
# Copying and viewing arrays
import numpy as np

original_array = np.array([1, 2, 3, 4, 5])
view = original_array[2:3]  # Create a view of the original array

view[0] = 99  # Modify the view

print("Original array after modifying the view:", original_array)

duplicate = original_array.copy()  # Create a copy of the original array

duplicate[0] = 100  # Modify the copy



print("Copy of the array:", duplicate)
print("Original array after modifying the copy:", original_array)

```

```python
numbers = np.array([1, 2, 3, 4, 5])

print(np.sqrt(numbers))
print(np.square(numbers))
print(np.abs(numbers))
print(np.exp(numbers))
print(np.log(numbers))
print(np.log10(numbers))
```

```python
print(np.sum(numbers))
print(np.mean(numbers))
print(np.median(numbers))
print(np.min(numbers))
print(np.max(numbers))
print(np.std(numbers))
print(np.var(numbers))
print(np.percentile(numbers, 25))
print(np.quantile(numbers, 0.25))
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers > 20)
print(np.any(numbers > 35))
print(np.all(numbers > 5))

condition_one = numbers >= 20
condition_two = numbers <= 30

print(np.logical_and(condition_one, condition_two))
print(np.logical_or(condition_one, condition_two))
print(np.logical_not(condition_one))
print(np.logical_xor(condition_one, condition_two))
```
